```python
X, Y = np.meshgrid(np.arange(10), np.arange(10)[::-1])
test_grid_thk = X + Y + 1
print(test_grid_thk)

max_layer_num = np.max(test_grid_thk)
values = np.zeros(max_layer_num)
counts = np.zeros(max_layer_num)

fig, ax = plt.subplots()
ax.imshow(test_grid_img_gray, cmap='gray')

for r in range(test_grid_thk.shape[0]):
    for c in range(test_grid_thk.shape[1]):
        layer_idx = test_grid_thk[r,c] - 1
        bounds = test_grid_px_idxs[r,c,:]

        ax.add_patch(patches.Rectangle((bounds[0], bounds[1]), width=bounds[2]-bounds[0], height=bounds[3]-bounds[1],
                                        linewidth=1, edgecolor='r', facecolor='none'))

        selection = test_grid_img_gray[bounds[1]:bounds[3], bounds[0]:bounds[2]]
        values[layer_idx] += np.median(selection)
        counts[layer_idx] += 1 

values = values / counts
plt.show()
        
```

```python
plt.plot(np.arange(max_layer_num)+1, values)

plt.xticks(np.arange(max_layer_num)+1)

plt.xlabel('Layer Count')
plt.ylabel('grayscale pixel value (8-bit)')
plt.title('Backlit brightness at 0.16mm layer height steps')

plt.show()
```
